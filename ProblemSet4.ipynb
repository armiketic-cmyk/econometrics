```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import scipy.stats as stats
```

### Question 3

```python
CPSData = pd.read_csv('cpsDATA.csv')
```

```python
X = sm.add_constant(CPSData['yrseduc'])
y = CPSData['ahe']
```

```python
model = sm.OLS(y, X).fit(cov_type='HC1')

print("=" * 60)
print("Part (a): OLS Regression with Heteroskedasticity-Robust SEs")
print("=" * 60)
print(model.summary())
```

```python
beta_educ  = model.params['yrseduc']
se_robust  = model.bse['yrseduc']
t_stat     = model.tvalues['yrseduc']
p_value    = model.pvalues['yrseduc']
n          = model.nobs
```

```python
print(f"\nCoefficient on yrseduc : {beta_educ:.4f}")
print(f"Robust SE              : {se_robust:.4f}")
```

```python
print("\n" + "=" * 60)
print("Part (b): Hypothesis Test  H0: β_educ = 0  vs  H1: β_educ ≠ 0")
print("=" * 60)
print(f"t-statistic : {t_stat:.4f}")
print(f"p-value     : {p_value:.6f}")
```

```python
alpha = 0.05
if p_value < alpha:
    print(f"→ Reject H0 at the {alpha*100:.0f}% level. "
          f"The education effect IS significantly different from zero.")
else:
    print(f"→ Fail to reject H0 at the {alpha*100:.0f}% level.")
```

```python
print("\n" + "=" * 60)
print("Part (c): 90% Confidence Interval for β_educ")
print("=" * 60)

df_resid  = model.df_resid
t_crit_90 = stats.t.ppf(0.95, df=df_resid)

ci_lower = beta_educ - t_crit_90 * se_robust
ci_upper = beta_educ + t_crit_90 * se_robust

print(f"t* (90%, df={int(df_resid)}) : {t_crit_90:.4f}")
print(f"90% CI : [{ci_lower:.4f},  {ci_upper:.4f}]")
print(f"\nInterpretation: We are 90% confident that an additional year")
print(f"of education raises average hourly earnings by between")
print(f"${ci_lower:.4f} and ${ci_upper:.4f} per hour.")

ci_sm = model.conf_int(alpha=0.10)
print(f"\n[Verification via statsmodels] 90% CI for yrseduc:")
print(f"  [{ci_sm.loc['yrseduc', 0]:.4f},  {ci_sm.loc['yrseduc', 1]:.4f}]")
```

### Question 4

```python
tr = pd.read_excel("TeachingRatings.xlsx", sheet_name = 0)
```

```python
X = sm.add_constant(tr['beauty'])
y = tr['course_eval']
```

```python
model = sm.OLS(y, X).fit(cov_type='HC1')

print("=" * 60)
print("Part (a): OLS Regression with Heteroskedasticity-Robust SEs")
print("=" * 60)
print(model.summary())
```

```python
beta_beauty = model.params['beauty']
se_robust   = model.bse['beauty']
t_stat      = model.tvalues['beauty']
p_value     = model.pvalues['beauty']
n           = model.nobs
```

```python
print(f"\nCoefficient on beauty : {beta_beauty:.4f}")
print(f"Robust SE             : {se_robust:.4f}")
```

```python
print("\n" + "=" * 60)
print("Part (b): Hypothesis Test  H0: β_beauty = 0  vs  H1: β_beauty ≠ 0")
print("=" * 60)
print(f"t-statistic : {t_stat:.4f}")
print(f"p-value     : {p_value:.6f}")
```

```python
alpha = 0.05
if p_value < alpha:
    print(f"→ Reject H0 at the {alpha*100:.0f}% level. "
          f"The beauty effect IS significantly different from zero.")
else:
    print(f"→ Fail to reject H0 at the {alpha*100:.0f}% level. "
          f"The beauty effect is NOT significantly different from zero.")
```

```python
print("\n" + "=" * 60)
print("Part (c): 99% Confidence Interval for β_beauty")
print("=" * 60)

df_resid  = model.df_resid
t_crit_99 = stats.t.ppf(0.995, df=df_resid)

ci_lower = beta_beauty - t_crit_99 * se_robust
ci_upper = beta_beauty + t_crit_99 * se_robust

print(f"t* (99%, df={int(df_resid)}) : {t_crit_99:.4f}")
print(f"99% CI : [{ci_lower:.4f},  {ci_upper:.4f}]")
print(f"\nInterpretation: We are 99% confident that a one-unit increase")
print(f"in professor beauty raises average course evaluations by between")
print(f"{ci_lower:.4f} and {ci_upper:.4f} points.")

# Verification via statsmodels
ci_sm = model.conf_int(alpha=0.01)
print(f"\n[Verification via statsmodels] 99% CI for beauty:")
print(f"  [{ci_sm.loc['beauty', 0]:.4f},  {ci_sm.loc['beauty', 1]:.4f}]")
```
